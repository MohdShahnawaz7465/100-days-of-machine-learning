# creating dataframe  for all pages

```python
final = pd.DataFrame()

for j in range(1, 4):   # pehle 3 pages se test karo, baad me range(1, 51) kar dena
    webpage = requests.get('https://books.toscrape.com/catalogue/page-{}.html'.format(j)).content
    soup = BeautifulSoup(webpage, 'lxml')
    company = soup.find_all('article', class_='product_pod')

    name = []
    rating = []
    price = []
    availability = []

    for i in company:
        try:
            name.append(i.find('h3').find('a')['title'])
        except:
            name.append(np.nan)

        try:
            rating.append(i.find('p', class_='star-rating')['class'][1])
        except:
            rating.append(np.nan)

        try:
            price.append(i.find('p', class_='price_color').text.strip())
        except:
            price.append(np.nan)

        try:
            availability.append(i.find('p', class_='instock').text.strip())
        except:
            availability.append(np.nan)

    df = pd.DataFrame({
        'name': name,
        'rating': rating,
        'price': price,
        'availability': availability
    })

    final = pd.concat([final, df], ignore_index=True)

final.shape
```
