# 2. Bar Plot (Numerical - Categorical)

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
sns.barplot(data=titanic, x='pclass', y='age', hue='sex')
plt.show()
```

# 3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(data=titanic, x='sex', y='age', hue='survived')
plt.show()
```

```python
sns.kdeplot(titanic[titanic['survived'] == 0]['age'], label='Not Survived')
sns.kdeplot(titanic[titanic['survived'] == 1]['age'], label='Survived')
plt.legend()
plt.show()
```

# 5. HeatMap (Categorical - Categorical)

```python
titanic.head(3)
```

```python
sns.heatmap(pd.crosstab(titanic['pclass'], titanic['survived']), annot=True, fmt='d')
plt.show()
```

```python
titanic.groupby('embarked')['survived'].mean() * 100
```

# 6. ClusterMap (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['parch'], titanic['survived']), annot=True, fmt='d')
plt.show()
```
