```python
frames = []
for page in range(1, 11):
    r = requests.get(
        'https://jsonplaceholder.typicode.com/posts',
        params={'_page': page, '_limit': 10},
        timeout=10,
    )
    r.raise_for_status()
    frames.append(pd.DataFrame(r.json())[['id', 'title', 'body']])

df = pd.concat(frames, ignore_index=True)
df.shape
```
